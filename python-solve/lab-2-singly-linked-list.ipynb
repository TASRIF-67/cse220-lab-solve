```python
#Run this cell
class Node:
  def __init__(self,elem,next = None):
    self.elem,self.next = elem,next

def createList(arr):
  head = Node(arr[0])
  tail = head
  for i in range(1,len(arr)):
    newNode = Node(arr[i])
    tail.next = newNode
    tail = newNode
  return head

def printLinkedList(head):
  temp = head
  while temp != None:
    if temp.next != None:
      print(temp.elem, end = '-->')
    else:
      print(temp.elem)
    temp = temp.next
  print()
```

### **Lab Task 1:** Assemble Conga Line
(no need to submit this)

```python
def assemble_conga_line(conga_line):
  temp = conga_line
  while (temp.next != None):
    if temp.elem > temp.next.elem:
      return False
    temp = temp.next

  return True

#[DO NOT MODIFY THE TESTER CODES BELOW]
#[THERE WILL BE 50% PENALTY IF IT'S MODIFIED]
print('==============Test Case 1=============')
conga_line = createList(np.array([10,15,34,41,56,72]))
print('Original Conga Line: ', end = ' ')
printLinkedList(conga_line)
returned_value = assemble_conga_line(conga_line)
print(returned_value) #This should print True
unittest.output_test(returned_value, True)
print()
print('==============Test Case 2=============')
conga_line = createList(np.array([10,15,44,41,56,72]))
print('Original Conga Line: ', end = ' ')
printLinkedList(conga_line)
returned_value = assemble_conga_line(conga_line)
print(returned_value) #This should print False
unittest.output_test(returned_value, False)
print()
```

### **Lab Task 2:** Word Decoder
(No need to submit this)

```python
def count_nodes(head):
  temp = head
  count = 0

  while (temp != None):
    count += 1
    temp = temp.next
  return count

def word_Decoder(head):
  num_nodes = count_nodes(head)
  idx = 13 % num_nodes
  curr_idx = 0
  temp = head

  new_head = None

  while (temp != None):
    if (curr_idx != 0 and curr_idx% idx == 0):
      new_node = Node(temp.elem)
      new_node.next = new_head
      new_head = new_node

    curr_idx += 1
    temp = temp.next

  dummy = Node(None)
  dummy.next = new_head
  return dummy



#Driver Code
print('==============Test Case 1=============')
head = createList(np.array(['B', 'M', 'D', 'T', 'N', 'O', 'A', 'P', 'S', 'C']))
print("Encoded Word:")
printLinkedList(head) #This should print B→M→D→T→N→O→A→P→S→C

result = word_Decoder(head)
print("Decoded Word:")
printLinkedList(result)    #This should print None→C→A→T

print('==============Test Case 2=============')

head = createList(np.array(['Z', 'O', 'T', 'N', 'X']))
print("Encoded Word:")
printLinkedList(head) #This should print Z→O→T→N→X

result = word_Decoder(head)
print("Decoded Word:")
printLinkedList(result)    #This should print None→N
```

### **Assignment Task 2:** Organize Books
(submit only the function. if extra helper functions were used. Submit those as well)

```python
def organizeBooks(head, arr):
  if head is None or head.next is None:
    return head

  curr1 = head
  i = 0

  #Selection sort on linked list
  while curr1 != None:
    max_node = curr1
    max_score = arr[i]
    max_idx = i

    curr2 = curr1.next
    j = i + 1

    while curr2 != None:
      if arr[j] > max_score:
        max_score = arr[j]
        max_node = curr2
        max_idx = j

      curr2 = curr2.next
      j += 1

    # Swap elems in linked list
    temp = curr1.elem
    curr1.elem = max_node.elem
    max_node.elem = temp

    # Swap elems in array
    temp_pop = arr[i]
    arr[i] = max_score
    arr[max_idx] = temp_pop

    curr1 = curr1.next
    i += 1

  return head



#[DO NOT MODIFY THE TESTER CODES BELOW]
#[THERE WILL BE 50% PENALTY IF IT'S MODIFIED]
print('==============Test Case 1=============')
books1 = createList(np.array(["Dune", "IT", "Coraline", "Inferno", "Twilight"]))
popularity1 = np.array([8, 10, 5, 10, 6])
print("Sample Given LinkedList and Popularity array:")
printLinkedList(books1)
print(popularity1)
print("\nExpected Output")
print("IT-->Inferno-->Dune-->Twilight-->Coraline")
print("Your Output")
books1 = organizeBooks(books1, popularity1)
printLinkedList(books1)

print('==============Test Case 2=============')
books2 = createList(np.array(["Hamlet", "Persuasion", "It", "Dracula", "Beloved"]))
popularity2 = np.array([7, 9, 9, 6, 7])
print("Sample Given LinkedList and Popularity array:")
printLinkedList(books2)
print(popularity2)
print("\nExpected Output")
print("Persuasion-->It-->Hamlet-->Beloved-->Dracula")
print("Your Output")
books2 = organizeBooks(books2, popularity2)
printLinkedList(books2)

print('==============Test Case 3=============')
books3 = createList(np.array(["Matilda", "Franny", "Foundation", "Carrie", "Misery"]))
popularity3 = np.array([5, 8, 8, 10, 6])
print("Sample Given LinkedList and Popularity array:")
printLinkedList(books3)
print(popularity3)
print("\nExpected Output")
print("Carrie-->Franny-->Foundation-->Misery-->Matilda")
print("Your Output")
books3 = organizeBooks(books3, popularity3)
printLinkedList(books3)
```

### **Assignment Task 3:** Alternative Merge
(submit only the function. if extra helper functions were used. Submit those as well)

```python
def alternate_merge(head1, head2):
  temp1 = head1
  temp2 = head2

  while (temp1 != None and temp2 != None):
    t1 = temp1.next
    t2 = temp2.next

    #Alternate connection
    temp1.next = temp2
    temp2.next = t1

    temp1 = t1
    temp2 = t2

  return head1



print('==============Test Case 1=============')
head1 = createList(np.array([1,2,6,8,11]))
head2 = createList(np.array([5,7,3,9,4]))

print("Linked List 1:")
printLinkedList(head1)
print("Linked List 2:")
printLinkedList(head2)

head = alternate_merge(head1, head2)
print("Merged Linked List:")
printLinkedList(head)    #This should print    1 --> 5 --> 2 --> 7 --> 6 --> 3 --> 8 --> 9 --> 11 --> 4


print('==============Test Case 2=============')
head1 = createList(np.array([5, 3, 2, -4]))
head2 = createList(np.array([-4, -6, 1]))

print("Linked List 1:")
printLinkedList(head1)
print("Linked List 2:")
printLinkedList(head2)

head = alternate_merge(head1, head2)
print("Merged Linked List:")
printLinkedList(head)    #This should print    5 → -4 -> 3 → -6 -> 2 -> 1 -> -4


print('==============Test Case 3=============')
head1 = createList(np.array([4, 2, -2, -4]))
head2 = createList(np.array([8, 6, 5, -3]))

print("Linked List 1:")
printLinkedList(head1)
print("Linked List 2:")
printLinkedList(head2)

head = alternate_merge(head1, head2)
print("Merged Linked List:")
printLinkedList(head)    #This should print   4-> 8 → 2-> 6 → -2 → 5 → -4 -> -3
```

### **Assignment Task 4:** ID Generator
(submit only the function. if extra helper function was used. Submit those as well)

```python
def idGenerator(head1, head2, head3):
  prev = None
  tail = head1
  curr = head1

  while curr != None:
    next_node = curr.next
    curr.next = prev
    prev = curr
    curr = next_node

  temp2 = head2
  temp3 = head3

  while (temp2 != None and temp3 != None):
    sum = temp2.elem + temp3.elem
    res = sum % 10

    new_node = Node(res)
    tail.next = new_node
    tail = new_node

    temp2 = temp2.next
    temp3 = temp3.next

  return prev


print('==============Test Case 1=============')
head1 = createList(np.array([0,3,2,2]))
head2 = createList(np.array([5,2,2,1]))
head3 = createList(np.array([4,3,2,1]))

print("Linked List 1:")
printLinkedList(head1)
print("Linked List 2:")
printLinkedList(head2)
print("Linked List 3:")
printLinkedList(head3)

result = idGenerator(head1, head2, head3)
print("New ID:")
printLinkedList(result)    #This should print  2 → 2 → 3 → 0 → 9 → 5 → 4 → 2


print('==============Test Case 2=============')
head1 = createList(np.array([0,3,9,1]))
head2 = createList(np.array([3,6,5,7]))
head3 = createList(np.array([2,4,3,8]))

print("Linked List 1:")
printLinkedList(head1)
print("Linked List 2:")
printLinkedList(head2)
print("Linked List 3:")
printLinkedList(head3)

result = idGenerator(head1, head2, head3)
print("New ID:")
printLinkedList(result)    #This should print 1 → 9 → 3 → 0 → 5 → 0 → 8 → 5
```
