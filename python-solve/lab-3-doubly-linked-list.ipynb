### Lab Task Class Design [NO NEED TO SUBMIT]

```python
class Patient:
    #write the constructor
    def __init__(self, id, name, age, bloodgroup, next, prev):
        self.id = id
        self.name = name
        self.age = age
        self.bloodgroup = bloodgroup
        self.next = next
        self.prev = prev
```

```python
class WRM:

    #here the constructor is already written for you
    def __init__(self):
        #Creating the dummy head, that's all this constructor will do
        self.dh = Patient(None,None,None,None,None,None)
        self.dh.next = self.dh
        self.dh.prev = self.dh

    def registerPatient(self,id, name, age, bloodgroup):
        new_patient = Patient(id, name, age, bloodgroup, None, None)

        tail = self.dh.prev

        tail.next = new_patient
        new_patient.prev = tail

        new_patient.next = self.dh
        self.dh.prev = new_patient

        print("Patient Registered!")

    def servePatient(self):
        if self.dh.next == self.dh:
          print("No Patient Waiting.")
          return None

        firstPatient = self.dh.next
        secondPatient = firstPatient.next

        self.dh.next = secondPatient
        secondPatient.prev = self.dh

        print("Serving Patient ID:", firstPatient.id)

    def showAllPatient(self):
        if self.dh.next == self.dh:
          print("No Patient Waiting.")
          return None

        temp = self.dh.next

        print("Waiting Patients: ")

        while temp != self.dh:
          print(temp.id, end = " ")
          temp = temp.next

        print()

    def canDoctorGoHome(self):
        if self.dh.next == self.dh:
          print("Yes. All patients are served.")
          return True

        print("No. Patients are waiting.")
        return False

    def cancelAll(self):
        self.dh.prev = self.dh
        self.dh.next = self.dh

        print("All appointments cancelled")

    def ReverseTheLine(self):
        if self.dh.next == self.dh:
            print("No patient waiting")
            return

        curr = self.dh

        while True:
          temp = curr.next

          curr.next = curr.prev
          curr.prev = temp
          curr = temp

          if curr == self.dh:
            break

        print("Line Reversed.")
```

```python
def sumOddAppend( head ):
  prev = head
  curr = head.next
  sum = 0

  while curr != head:
    if curr.elem % 2 != 0:
      sum += curr.elem
      prev.next = curr.next
    else:
      prev = curr

    curr = curr.next

  new_node = DNode(sum)
  new_node.next = head
  prev.next= new_node

  return head

```

### Assignment Task 6 [Must Submit]

```python
def pairJoin(dhead1, dhead2):
    if dhead1.next is None or dhead1.next == dhead1:
        return dhead1

    curr1 = dhead1.next
    curr2 = dhead2.next
    last = None

    while curr1 is not None:
        next1 = curr1.next
        next2 = curr2.next

        curr1.next = curr2
        curr2.prev = curr1
        curr2.next = next1
        if next1 is not None:
            next1.prev = curr2

        last = curr2  #track last inserted node
        curr1 = next1
        curr2 = next2

    # make it circular
    last.next = dhead1
    dhead1.prev = last

    return dhead1
```

### Assignment Task 7 [Must Submit]

```python
def rangeMove(dhead1, start, end):
    range_head = None
    range_tail = None

    curr = dhead1.next

    while curr != dhead1:
        next_node = curr.next

        if start <= curr.elem <= end:
            # detach curr from the main list
            curr.prev.next = curr.next
            curr.next.prev = curr.prev

            # append curr to the range chain
            if range_head is None:
                range_head = curr
                range_tail = curr
                curr.next = None
                curr.prev = None
            else:
                range_tail.next = curr
                curr.prev = range_tail
                curr.next = None
                range_tail = curr

        curr = next_node

    if range_head is None:
        return None

    # splice range chain at the back
    last_node = dhead1.prev

    last_node.next = range_head
    range_head.prev = last_node

    range_tail.next = dhead1
    dhead1.prev = range_tail

    return dhead1
```
